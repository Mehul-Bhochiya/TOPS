Healthcare Industry
Design a Python class ClinicAppointment that manages patient appointments in a clinic.
The system should have the following features:
➔ Book Appointment:
● Prompt for patient name, age, mobile number, and preferred doctor.
● Show time slots (10am, 11am, 12pm, 2pm, 3pm).
● Check slot availability and confirm booking.
➔ View/Cancel Appointment:
● Allow patient to view or cancel their appointment using mobile number.
➔ Doctor Availability:
● Maintain a maximum of 3 appointments per time slot per doctor.
➔ Data Persistence:
● Store appointments in memory only (no files/dbs required).

```python
class ClinicAppointment:
    def __init__(self):
        self.appointments = {}
        self.time_slots   = ["10am", "11am", "12pm", "2pm", "3pm"]
        self.doctor_names = ["Dr. Shah", "Dr. Pandya", "Dr. Patel"]
        self.max_per_slot = 3

    def book_appointment(self):
        name   = input("Enter patient name: ").strip()
        age    = int(input("Enter age: "))
        mobile = input("Enter mobile number: ").strip()

        print("\nAvailable doctors:")
        for i, doc in enumerate(self.doctor_names, 1):
            print(f"  {i}. {doc}")
        doctor = self.doctor_names[int(input("Choose doctor (number): ")) - 1]

        print("\nAvailable time slots:")
        for i, slot in enumerate(self.time_slots, 1):
            print(f"  {i}. {slot}")
        slot = self.time_slots[int(input("Choose time slot (number): ")) - 1]

        key = (doctor, slot)
        if key not in self.appointments:
            self.appointments[key] = []

        if len(self.appointments[key]) >= self.max_per_slot:
            print("Slot is full. Please choose another slot.")
            return

        self.appointments[key].append({
            "name":        name,
            "age":         age,
            "mobile":      mobile,
            "doctor_name": doctor,  
            "time_slot":   slot      
        })
        print(f"Appointment booked with {doctor} at {slot}.")

    def view_appointments(self):
        mobile = input("Enter mobile number: ").strip()
        found  = False

        for (doctor, slot), patients in self.appointments.items():
            for patient in patients:
                if patient["mobile"] == mobile:
                    print(f"\n  Name        : {patient['name']}")
                    print(f"  Age         : {patient['age']}")
                    print(f"  Mobile      : {patient['mobile']}")
                    print(f"  Doctor      : {patient['doctor_name']}")
                    print(f"  Time Slot   : {patient['time_slot']}")
                    found = True

        if not found:
            print("No appointment found for this mobile number.")

    def cancel_appointment(self):
        mobile = input("Enter mobile number: ").strip()
        found  = False

        for key in list(self.appointments.keys()):
            patients = self.appointments[key]
            for patient in patients:
                if patient["mobile"] == mobile:
                    patients.remove(patient)
                    print("Appointment cancelled successfully.")
                    found = True
                    break  
            if not self.appointments[key]:
                del self.appointments[key]
            if found:
                break       

        if not found:
            print("No appointment found for this mobile number.")

    def run(self):
        while True:
            print("\n===== Clinic Appointment System =====")
            print("1. Book Appointment")
            print("2. View Appointment")
            print("3. Cancel Appointment")
            print("4. Exit")

            choice = input("Enter choice: ").strip()

            if choice == "1":
                self.book_appointment()
            elif choice == "2":
                self.view_appointments()
            elif choice == "3":
                self.cancel_appointment()
            elif choice == "4":
                print("Exiting. Goodbye!")
                break
            else:
                print("Invalid choice. Please enter 1–4.")
```

School Management System
Design a Python class SchoolManagement that helps manage student admissions and
records. The system should support:
➔ New Admission:
● Collect student name, age, class (1–12), and guardian's mobile number.
● Assign a unique student ID automatically.
● Validate age: must be between 5 and 18.
● Validate mobile number: must be 10 digits.
➔ View Student Details:
● Allow lookup using student ID.
➔ Update Student Info:
● Update mobile number or class.
➔ Remove Student Record:
● Remove a student using their student ID.
➔ Exit System

```python
class SchoolManagement:
    def __init__(self):
        self.students   = {}   
        self._id_counter = 1   

    def _generate_id(self):
        student_id = f"STU{self._id_counter:04d}"  
        self._id_counter += 1
        return student_id

    def new_admission(self):
        name = input("Enter student name: ").strip()


        while True:
            try:
                age = int(input("Enter age (5–18): "))
                if 5 <= age <= 18:
                    break
                print("Age must be between 5 and 18.")
            except ValueError:
                print("Please enter a valid number.")


        while True:
            try:
                student_class = int(input("Enter class (1–12): "))
                if 1 <= student_class <= 12:
                    break
                print("Class must be between 1 and 12.")
            except ValueError:
                print("Please enter a valid number.")


        while True:
            mobile = input("Enter guardian's mobile number (10 digits): ").strip()
            if mobile.isdigit() and len(mobile) == 10:
                break
            print("Mobile number must be exactly 10 digits.")

        student_id = self._generate_id()

        self.students[student_id] = {
            "name":          name,
            "age":           age,
            "class":         student_class,
            "mobile":        mobile,
            "student_id":    student_id
        }

        print(f"\nAdmission successful! Student ID: {student_id}")

    def view_student(self):
        student_id = input("Enter Student ID: ").strip().upper()

        if student_id not in self.students:
            print("Student not found.")
            return

        s = self.students[student_id]
        print(f"\n  Student ID : {s['student_id']}")
        print(f"  Name       : {s['name']}")
        print(f"  Age        : {s['age']}")
        print(f"  Class      : {s['class']}")
        print(f"  Mobile     : {s['mobile']}")

    def update_student(self):
        student_id = input("Enter Student ID: ").strip().upper()

        if student_id not in self.students:
            print("Student not found.")
            return

        print("What do you want to update?")
        print("1. Mobile Number")
        print("2. Class")
        choice = input("Enter choice: ").strip()

        if choice == "1":
            while True:
                mobile = input("Enter new mobile number (10 digits): ").strip()
                if mobile.isdigit() and len(mobile) == 10:
                    break
                print("Mobile number must be exactly 10 digits.")
            self.students[student_id]["mobile"] = mobile
            print("Mobile number updated.")

        elif choice == "2":
            while True:
                try:
                    new_class = int(input("Enter new class (1–12): "))
                    if 1 <= new_class <= 12:
                        break
                    print("Class must be between 1 and 12.")
                except ValueError:
                    print("Please enter a valid number.")
            self.students[student_id]["class"] = new_class
            print("Class updated.")

        else:
            print("Invalid choice.")

    def remove_student(self):
        student_id = input("Enter Student ID to remove: ").strip().upper()

        if student_id not in self.students:
            print("Student not found.")
            return

        del self.students[student_id]
        print(f"Student {student_id} removed successfully.")

    def run(self):
        while True:
            print("\n===== School Management System =====")
            print("1. New Admission")
            print("2. View Student Details")
            print("3. Update Student Info")
            print("4. Remove Student Record")
            print("5. Exit")

            choice = input("Enter choice: ").strip()

            if choice == "1":
                self.new_admission()
            elif choice == "2":
                self.view_student()
            elif choice == "3":
                self.update_student()
            elif choice == "4":
                self.remove_student()
            elif choice == "5":
                print("Exiting School Management System. Goodbye!")
                break
            else:
                print("Invalid choice. Please enter 1–5.")
```

Transport Reservation System (Bus Ticketing)
Design a Python class BusReservation that simulates a basic bus ticket booking system.
Features should include:
➔ Show Available Routes:
● Predefined city routes with fixed prices.
● Example: "Mumbai to Pune - ₹500", "Delhi to Jaipur - ₹600", etc.
➔ Book Ticket:
● Enter passenger name, age, mobile, and route.
● Assign seat number (max 40 per bus per route).
● Generate a unique ticket ID.
➔ View Ticket:
● Lookup using ticket ID.
➔ Cancel Ticket:
● Cancel the ticket if it exists.
➔ Exit

```python
class BusReservation:
    def __init__(self):
        self.routes = {
            "Mumbai to Pune":      500,
            "Delhi to Jaipur":     600,
            "Ahmedabad to Surat":  300,
            "Bangalore to Mysore": 250,
            "Chennai to Pondicherry": 200
        }

        self.max_seats     = 40
        self.tickets       = {}   
        self.seat_counter  = {}   
        self._ticket_seed  = 1    

    def _generate_ticket_id(self):
        ticket_id = f"TKT{self._ticket_seed:05d}"   
        self._ticket_seed += 1
        return ticket_id

    def show_routes(self):
        print("\n===== Available Routes =====")
        for i, (route, price) in enumerate(self.routes.items(), 1):
            booked = self.seat_counter.get(route, 0)
            seats_left = self.max_seats - booked
            print(f"  {i}. {route} — ₹{price}  |  Seats available: {seats_left}")

    def book_ticket(self):
        name = input("Enter passenger name: ").strip()
        age  = int(input("Enter age: "))

        while True:
            mobile = input("Enter mobile number (10 digits): ").strip()
            if mobile.isdigit() and len(mobile) == 10:
                break
            print("Mobile number must be exactly 10 digits.")

        self.show_routes()
        route_list = list(self.routes.keys())
        while True:
            try:
                route_choice = int(input("Choose route (number): ")) - 1
                if 0 <= route_choice < len(route_list):
                    break
                print(f"Please enter a number between 1 and {len(route_list)}.")
            except ValueError:
                print("Please enter a valid number.")

        selected_route = route_list[route_choice]
        booked_seats   = self.seat_counter.get(selected_route, 0)

        if booked_seats >= self.max_seats:
            print("Sorry, this route is fully booked.")
            return

        seat_number = booked_seats + 1
        self.seat_counter[selected_route] = seat_number

        ticket_id = self._generate_ticket_id()
        price     = self.routes[selected_route]

        self.tickets[ticket_id] = {
            "ticket_id":  ticket_id,
            "name":       name,
            "age":        age,
            "mobile":     mobile,
            "route":      selected_route,
            "seat":       seat_number,
            "price":      price
        }

        print(f"\nTicket booked successfully!")
        print(f"  Ticket ID   : {ticket_id}")
        print(f"  Route       : {selected_route}")
        print(f"  Seat Number : {seat_number}")
        print(f"  Fare        : ₹{price}")

    def view_ticket(self):
        ticket_id = input("Enter Ticket ID: ").strip().upper()

        if ticket_id not in self.tickets:
            print("Ticket not found.")
            return

        t = self.tickets[ticket_id]
        print(f"\n  Ticket ID   : {t['ticket_id']}")
        print(f"  Name        : {t['name']}")
        print(f"  Age         : {t['age']}")
        print(f"  Mobile      : {t['mobile']}")
        print(f"  Route       : {t['route']}")
        print(f"  Seat Number : {t['seat']}")
        print(f"  Fare        : ₹{t['price']}")

    def cancel_ticket(self):
        ticket_id = input("Enter Ticket ID to cancel: ").strip().upper()

        if ticket_id not in self.tickets:
            print("Ticket not found.")
            return

        route = self.tickets[ticket_id]["route"]


        if self.seat_counter.get(route, 0) > 0:
            self.seat_counter[route] -= 1

        del self.tickets[ticket_id]
        print(f"Ticket {ticket_id} cancelled successfully.")

    def run(self):
        while True:
            print("\n===== Bus Reservation System =====")
            print("1. Show Available Routes")
            print("2. Book Ticket")
            print("3. View Ticket")
            print("4. Cancel Ticket")
            print("5. Exit")

            choice = input("Enter choice: ").strip()

            if choice == "1":
                self.show_routes()
            elif choice == "2":
                self.book_ticket()
            elif choice == "3":
                self.view_ticket()
            elif choice == "4":
                self.cancel_ticket()
            elif choice == "5":
                print("Exiting Bus Reservation System. Goodbye!")
                break
            else:
                print("Invalid choice. Please enter 1–5.")


bus = BusReservation()
bus.run()
```
