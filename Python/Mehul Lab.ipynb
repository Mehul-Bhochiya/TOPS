Mini Project 1: Sales Calculator Module
Create functions:
•	calculate_revenue() 
•	calculate_discount() 
•	calculate_tax() 
•	final_price() 
Use them together in a billing flow.

```python
def calculate_revenue(price, qty):
    return price * qty


def calculate_discount(amount, discount_per):
    return amount * discount_per / 100


def calculate_tax(amount, tax_per):
    return amount * tax_per / 100


def final_price(amount, discount, tax):
    return amount - discount + tax

price = float(input("enter price: "))
qty = int(input("enter quantity: "))
discount_per = float(input("enter discount %: "))
tax_per = float(input("enter tax %: "))

revenue = calculate_revenue(price, qty)
discount = calculate_discount(revenue, discount_per)
tax = calculate_tax(revenue, tax_per)
final = final_price(revenue, discount, tax)

print("Revenue:", revenue)
print("Discount:", discount)
print("Tax:", tax)
print("Final Price:", final)
```

Mini Project 2: Log Analyzer
Input:
•	log file 
Tasks:
•	count ERROR / INFO / WARNING 
•	extract timestamps 
•	generate summary report 

```python
import re

def analyze_log(file_path):
    try:
        with open(r"C:\Users\Mehul\OneDrive\Documents\TOPS\Python\log.txt", 'r') as file:
            logs = file.readlines()

        error_count = 0
        info_count = 0
        warning_count = 0
        timestamps = []

        timestamp_pattern = r'\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}'

        for line in logs:
            if "ERROR" in line:
                error_count += 1
            elif "INFO" in line:
                info_count += 1
            elif "WARNING" in line:
                warning_count += 1

            match = re.search(timestamp_pattern, line)
            if match:
                timestamps.append(match.group())

        print("===== LOG SUMMARY REPORT =====")
        print(f"ERROR: {error_count}")
        print(f"INFO: {info_count}")
        print(f"WARNING: {warning_count}")
        print("\nTimestamps:")
        for t in timestamps:
            print(t)

    except FileNotFoundError:
        print("Log file not found!")

analyze_log("log.txt")
```

Mini Project 3: Data Validation Engine
Validate:
•	email 
•	phone 
•	date 
•	PIN code 
Return:
•	valid / invalid status

```python
import re

def validate_email(email):
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    return bool(re.match(pattern, email))


def validate_phone(phone):
    pattern = r'^\d{10}$'
    return bool(re.match(pattern, phone))


def validate_date(date):
    # Format: YYYY-MM-DD
    pattern = r'^\d{4}-\d{2}-\d{2}$'
    return bool(re.match(pattern, date))


def validate_pin(pin):
    pattern = r'^\d{6}$'
    return bool(re.match(pattern, pin))


# Test
print("Email:", validate_email("test@gmail.com"))
print("Phone:", validate_phone("9876543210"))
print("Date:", validate_date("2025-01-01"))
print("PIN:", validate_pin("380001"))
```

```python
# Create abstract class
# Implement abstract method
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w


r = Rectangle(5, 4)
print(r.area())
```

```python
# Create customer credit system
class Customer:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def add_credit(self, amount):
        self.__balance += amount

    def use_credit(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    def show_balance(self):
        print(self.__balance)

c1 = Customer("Mehul", 1000)
c1.add_credit(500)
c1.use_credit(300)
c1.show_balance()
```

Mini Project 2: Model Configuration Manager
•	Encapsulated parameters 
•	Abstract training interface

```python
from abc import ABC, abstractmethod

# Encapsulated config
class ModelConfig:
    def __init__(self, lr, epochs):
        self.__learning_rate = lr
        self.__epochs = epochs

    def get_config(self):
        return self.__learning_rate, self.__epochs


# Abstract trainer
class Trainer(ABC):
    @abstractmethod
    def train(self):
        pass


# Concrete implementation
class LinearModelTrainer(Trainer):
    def __init__(self, config: ModelConfig):
        self.config = config

    def train(self):
        lr, epochs = self.config.get_config()
        print(f"Training model with lr={lr}, epochs={epochs}")


# usage
config = ModelConfig(0.01, 10)
trainer = LinearModelTrainer(config)
trainer.train()
```

Mini Project 1: Employee Bonus System
Classes:
•	Employee 
•	Manager 
•	Director 
Each has:
•	different bonus logic 
•	common salary structure 

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        return self.salary * 0.10   # default 10%

    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Bonus: {self.calculate_bonus()}")
        print("-" * 30)


class Manager(Employee):
    def calculate_bonus(self):
        return self.salary * 0.20   # 20% bonus


class Director(Employee):
    def calculate_bonus(self):
        return self.salary * 0.30   # 30% bonus


employees = [
    Employee("Mehul", 50000),
    Manager("Raj", 80000),
    Director("Anita", 120000)
]

for emp in employees:
    emp.show_details()
```
