Mini Project 1 — Customer Data Cleaning 
Dataset [100,200,np.nan,300,400,np.inf,5000] 
Tasks 1 Remove NaN and infinite values 
2 Clip values between 0–1000 
3 Calculate average

```python
arr1 = np.array([100, 200, np.nan, 300, 400, np.inf, 5000])

clean_arr = arr1[np.isfinite(arr1)]
clipped_arr = np.clip(clean_arr, 0, 1000)
average = np.mean(clipped_arr)

print("Original Array:", arr1)
print("Cleaned Array:", clean_arr)
print("Clipped Array:", clipped_arr)
print("Average:", average)
```
