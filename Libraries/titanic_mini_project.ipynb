```python
df  = df.drop_duplicates().reset_index(drop=True)
```

```python
df.describe()
```

```python
df.age = df.age.fillna(df.age.mean())
```

```python
df.isnull().sum()
```

```python
df
```

```python
df.embark_town.value_counts()
```

```python
df.embarked.value_counts()
```

```python
df.embarked = df.embarked.fillna("S")
```

```python
df.embark_town = df.embark_town.fillna("Southampton")
```

```python
df.isnull().sum()
```

```python
# Random filling

df.deck = df.deck.bfill()
```

```python
df.deck = df.deck.ffill()
```

```python
df.isnull().sum()
```

```python
pd.DataFrame(df.embark_town.value_counts()).rename(columns= {"count": "city"}).plot.bar() # .rename --- rename the count in output
```

```python
pd.DataFrame(df.embark_town.value_counts()).rename(columns= {"count": "city"}).T.plot.bar() # .T  ---  is transpose convert column in row or row into column
```

```python
sns.catplot(df, x = "embark_town", y = "age", kind="bar", errorbar=None, estimator= "count") # y is not necessary in this chart
```

```python
pd.DataFrame(df.alone.value_counts()).rename(columns= {"count": "city"}).plot.bar()
```

```python
pd.DataFrame(df.who.value_counts()).rename(columns= {"count": "city"}).plot.bar()
```

```python
pd.DataFrame(df.groupby(["alive","alone"], observed=False)["alone"].count()).plot.bar()
```
